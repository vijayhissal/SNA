```r
library(dplyr)
library(stringr)

set.seed(456)

# Create self-contained senator_tweet_sample dataset
senator_tweet_sample <- data.frame(
  screen_name = c(
    "SenSmith", "SenSmith", "SenJohnson", "SenJohnson", "SenWilliams",
    "SenWilliams", "SenBrown", "SenBrown", "SenJones", "SenJones",
    "SenGarcia", "SenGarcia", "SenMiller", "SenMiller", "SenDavis", "SenDavis"
  ),
  gender = c(
    "F", "F", "M", "M", "F",
    "F", "M", "M", "F", "F",
    "M", "M", "M", "M", "F", "F"
  ),
  text = c(
    "Great meeting today with @SenJohnson and @DeptOfEnergy on clean grid funding!",
    "Thank you @LocalNews for covering our infrastructure town hall.",
    "Working across the aisle with @SenSmith @SenBrown and @SenateCommerce.",
    "Proud to support our local farmers across the state this week.",
    "Joined @SenJones @SenDavis @EPA and @USDA to discuss rural water grants!",
    "Tune in live at 3 PM for our Judiciary Committee hearing.",
    "Honored to speak alongside @SenMiller at the veterans memorial.",
    "New bipartisan bill introduced today to lower prescription drug costs.",
    "Thank you @SenWilliams and @MayorOffice for hosting today's roundtable!",
    "Congratulations to @StateUniversity graduates on their commencement!",
    "Discussing semiconductor manufacturing with @CommerceGov and @SenMiller.",
    "Wishing everyone a safe and happy weekend back home.",
    "Productive briefing with @DOD and @SenGarcia on cybersecurity readiness.",
    "Our office is open to help constituents with federal agency casework.",
    "Thrilled to partner with @SenWilliams @SenSmith @NIH on biomedical research!",
    "Met with student leaders from @HighSchoolRobotics today—inspiring work!"
  ),
  favorite_count = c(
    420, 210, 580, 140, 710,
    185, 310, 160, 490, 275,
    390, 120, 350, 150, 680, 260
  ),
  stringsAsFactors = FALSE
)

head(senator_tweet_sample)
```

```r
# Count "@" followed by non-whitespace characters ("@\\S+") in each tweet
senator_tweet_sample <- senator_tweet_sample %>%
  mutate(mention_count = str_count(text, "@\\S+"))

# Preview tweet text alongside extracted mention count and favorites
senator_tweet_sample %>%
  select(screen_name, mention_count, favorite_count) %>%
  head(8)

# Compute Pearson correlation between mention_count and favorite_count
mention_fav_cor <- cor(
  senator_tweet_sample$mention_count,
  senator_tweet_sample$favorite_count,
  use = "complete.obs"
)

cat("Correlation between number of mentions and favorites:", round(mention_fav_cor, 4), "\n\n")

# Run full significance test
cor.test(senator_tweet_sample$mention_count, senator_tweet_sample$favorite_count)
```

```r
avg_mentions_by_gender <- senator_tweet_sample %>%
  group_by(gender) %>%
  summarise(
    total_tweets = n(),
    total_mentions = sum(mention_count),
    avg_mentions_per_tweet = mean(mention_count, na.rm = TRUE)
  )

print(avg_mentions_by_gender)
```
